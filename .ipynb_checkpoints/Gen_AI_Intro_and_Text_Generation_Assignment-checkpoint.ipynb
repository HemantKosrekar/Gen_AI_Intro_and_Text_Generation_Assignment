### 1. What is Generative AI?

**Generative AI (GenAI)** is a subset of artificial intelligence capable of creating new, original content—such as text, images, audio, video, and code—that resembles human-created data. Unlike traditional models that only classify or predict labels from existing inputs, generative models learn the underlying probability distribution and structural patterns of training data to synthesize novel outputs.

### 2. How is Generative AI different from traditional AI?

- **Objective:** Traditional AI (discriminative AI) focuses on pattern recognition, classification, regression, and decision-making based on explicit rules or labeled boundaries ($P(Y|X)$). Generative AI focuses on learning the joint or data distribution ($P(X)$ or $P(X,Y)$) to generate new samples.
- **Output:** Traditional AI outputs predictions, scores, or categories (e.g., spam vs. not spam). Generative AI outputs rich, unstructured artifacts (e.g., essays, photorealistic images, software code).
- **Data & Architecture:** Generative AI relies heavily on deep foundation models (Transformers, Diffusion models, VAEs, GANs) trained via self-supervised learning on massive datasets.

### 3. Name two applications of Generative AI in the industry.

1. **Software Development & Code Generation:** Automated code completion, bug fixing, and unit test generation (e.g., GitHub Copilot, Claude Code).
2. **Healthcare & Drug Discovery:** Designing novel molecular structures, predicting protein folding, and summarizing clinical notes.

### 4. What are some challenges associated with Generative AI?

- **Hallucinations:** Generating plausible-sounding yet factually incorrect or fabricated information.
- **High Computational & Energy Costs:** Training and serving large foundation models require massive GPU clusters and power.
- **Bias & Toxicity:** Models can inherit and amplify social biases, stereotypes, or harmful language present in web-scale training corpora.
- **Intellectual Property & Privacy:** Risks of memorizing and reproducing copyrighted material or sensitive personal data.

### 5. Why is Generative AI important for modern applications?

Generative AI transforms human-computer interaction by enabling natural language interfaces, automating complex creative and cognitive workflows, accelerating R&D cycles, and hyper-personalizing user experiences at scale across industries like education, finance, media, and engineering.

### 6. What is probabilistic modeling in the context of Generative AI?

**Probabilistic modeling** represents data generation as a statistical process by estimating the underlying probability distribution $P(X)$ of the training dataset. By modeling uncertainties and token/feature likelihoods, a generative model can sample new data points from high-probability regions of that learned distribution.

### 7. Define a generative model.

A **generative model** is a statistical or machine learning model that learns the true data distribution $P(X)$ (or joint distribution $P(X, Y)$) of a dataset so that it can draw new samples $\hat{x} \sim P_{\theta}(X)$ that are statistically consistent with the training data.

### 8. Explain how an n-gram model works in text generation.

An **n-gram model** predicts the next word in a sequence based on the immediately preceding $n-1$ words using conditional probabilities estimated from corpus frequency counts:
$$P(w_t \mid w_{t-n+1}, \dots, w_{t-1}) = \frac{\text{Count}(w_{t-n+1}, \dots, w_{t-1}, w_t)}{\text{Count}(w_{t-n+1}, \dots, w_{t-1})}$$
During text generation, the model starts with an initial context of $n-1$ tokens, looks up the probability distribution over all possible next words, samples (or picks the highest-probability) next word, appends it to the sequence, and repeats.

### 9. What are the limitations of n-gram models?

- **Data Sparsity:** Many valid word combinations never appear in the training corpus, resulting in zero probabilities.
- **Curse of Dimensionality:** As $n$ increases, the number of possible n-grams grows exponentially ($|V|^n$), requiring infeasible storage and data.
- **Short Context Window:** They cannot capture long-range dependencies beyond $n-1$ words.
- **Lack of Semantic Generalization:** Words are treated as discrete symbols without understanding synonyms or meaning (e.g., *cat* and *feline* share no statistical strength).

### 10. How can you improve the performance of an n-gram model?

- **Smoothing Techniques:** Apply Laplace (add-one), Good-Turing, or Kneser-Ney smoothing to assign non-zero probabilities to unseen n-grams.
- **Backoff and Interpolation:** Fall back to lower-order n-grams (e.g., bigrams or unigrams) when higher-order n-gram counts are sparse, or combine probabilities across multiple orders.
- **Larger & Cleaner Corpora:** Increase training data coverage and apply subword tokenization or lowercasing/normalization.

### 11. What is the Markov assumption, and how does it apply to text generation?

The **Markov assumption** states that the probability of a future state depends only on a fixed number of immediately preceding states rather than the entire history:
$$P(w_t \mid w_1, w_2, \dots, w_{t-1}) \approx P(w_t \mid w_{t-k}, \dots, w_{t-1})$$
In text generation, this simplifies language modeling by allowing a model to generate the next word using only a local window of $k$ prior words instead of conditioning on every word since the beginning of the document.

### 12. Why are probabilistic models important in generative AI?

Probabilistic models provide a principled mathematical framework to handle ambiguity and variation in real-world data. Because there are many valid continuations for a prompt or multiple valid images for a description, modeling a probability distribution allows generative AI to produce diverse, creative, and realistic outputs rather than rigid, deterministic responses.

### 13. What is an autoencoder?

An **autoencoder** is an unsupervised neural network trained to reconstruct its input at the output layer. It consists of two parts:
1. **Encoder:** Compresses high-dimensional input $x$ into a lower-dimensional latent representation (bottleneck) $z = f(x)$.
2. **Decoder:** Reconstructs the original input $\hat{x} = g(z)$ from the latent vector $z$ by minimizing reconstruction loss (e.g., MSE or cross-entropy).

### 14. How does a VAE differ from a standard autoencoder?

- **Latent Representation:** A standard autoencoder maps an input to a fixed, deterministic point $z$ in latent space. A **Variational Autoencoder (VAE)** maps an input to parameters of a probability distribution—mean $\mu$ and variance $\sigma^2$—and samples $z \sim \mathcal{N}(\mu, \sigma^2)$ using the reparameterization trick.
- **Regularized Latent Space:** VAEs include a KL divergence loss term that forces the latent space to be continuous and smooth, whereas standard autoencoders often have discontinuous, fragmented latent spaces unsuitable for random sampling.

### 15. Why are VAEs useful in generative modeling?

Because a VAE enforces a smooth, continuous latent space aligned with a standard normal prior $\mathcal{N}(0, I)$, any vector $z$ sampled from that prior can be passed through the decoder to generate coherent, novel, and realistic data points—as well as smooth interpolations between existing examples.

### 16. What role does the decoder play in an autoencoder?

The **decoder** acts as the generative reconstruction engine. It takes the compressed latent representation $z$ from the bottleneck layer and maps it back to the original high-dimensional input space ($\hat{x} = g(z)$), learning to expand abstract features into structured data.

### 17. How does the latent space affect text generation in a VAE?

- **Smoothness & Continuity:** A well-regularized latent space ensures that nearby latent vectors decode into semantically and grammatically similar sentences.
- **Diversity vs. Posterior Collapse:** A well-structured latent space allows controlled sampling and attribute interpolation (e.g., changing sentiment or tense). However, if the decoder is too powerful (like an autoregressive RNN/Transformer), *posterior collapse* can occur, causing the decoder to ignore the latent variable $z$.

### 18. What is the purpose of the Kullback-Leibler (KL) divergence term in VAEs?

The **KL divergence** term $D_{KL}(q_{\phi}(z|x) \parallel p(z))$ measures how much the learned encoder distribution $q_{\phi}(z|x)$ diverges from a chosen prior distribution (typically $\mathcal{N}(0, I)$). It acts as a regularizer that prevents the encoder from memorizing inputs into isolated delta spikes, ensuring a continuous, dense latent space from which new samples can be reliably drawn.

### 19. How can you prevent overfitting in a VAE?

- **KL Annealing (Warm-up):** Gradually increase the weight $\beta$ of the KL divergence term during early training epochs.
- **Dropout & Weight Decay (L2 Regularization):** Apply dropout layers and L2 penalties in both encoder and decoder networks.
- **Word Dropout (for text VAEs):** Randomly mask input tokens fed into the decoder so it relies on the latent code $z$.
- **Early Stopping & Data Augmentation:** Stop training when validation ELBO plateaus and expand training data diversity.

### 20. Explain why VAEs are commonly used for unsupervised learning tasks.

VAEs require **no manual class labels**—the input data itself serves as the supervision target via reconstruction loss. By learning compact, disentangled latent representations directly from raw unlabeled data, VAEs excel at representation learning, anomaly detection, clustering, and synthetic data generation.

### 21. What is a transformer model?

A **Transformer** (introduced by Vaswani et al., 2017) is a deep learning architecture designed for sequence modeling that relies entirely on **self-attention mechanisms** rather than recurrence (RNNs) or convolutions. It processes all tokens in parallel using multi-head attention, positional encodings, feed-forward layers, residual connections, and layer normalization.

### 22. Explain the purpose of self-attention in transformers.

**Self-attention** allows every token in a sequence to dynamically weigh and attend to all other tokens in the same sequence, regardless of their distance. Using Query ($Q$), Key ($K$), and Value ($V$) projections:
$$\text{Attention}(Q, K, V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right)V$$
it resolves contextual relationships (such as coreference, syntax, and long-range semantics) in a single step.

### 23. How does a GPT model generate text?

GPT (Generative Pre-trained Transformer) is an **autoregressive, decoder-only Transformer** trained with causal (masked) self-attention so each token can only attend to preceding tokens. Given a prompt, GPT computes a probability distribution over the vocabulary for the next token, selects a token using a decoding strategy (greedy, beam search, top-$k$, top-$p$, or temperature sampling), appends it to the sequence, and repeats iteratively until an end-of-sequence token or length limit is reached.

### 24. What are the key differences between a GPT model and an RNN?

| Feature | GPT (Decoder-Only Transformer) | Recurrent Neural Network (RNN) |
| :--- | :--- | :--- |
| **Processing** | Parallel across all tokens during training | Sequential step-by-step ($t=1 \dots T$) |
| **Long-Range Dependencies** | Direct $O(1)$ path via self-attention | Degrades over distance due to vanishing/exploding gradients |
| **Scalability** | Scales efficiently to billions of parameters | Hard to parallelize; slow on long sequences |
| **Context Representation** | Positional embeddings + multi-head attention | Compressed into a single fixed-size hidden state vector $h_t$ |

### 25. How does fine-tuning improve a pre-trained GPT model?

Pre-training teaches GPT general grammar, syntax, and broad world knowledge. **Fine-tuning** updates the model weights (or lightweight adapter layers like LoRA) on a smaller, curated task- or domain-specific dataset, aligning the model's output style, formatting, terminology, and behavior with downstream requirements (e.g., customer support, medical summarization, or instruction following).

### 26. What is zero-shot learning in the context of GPT models?

**Zero-shot learning** is the ability of a pre-trained GPT model to perform a task it was never explicitly fine-tuned on, solely by understanding natural language instructions in the prompt (e.g., *"Translate the following English sentence to French:"* without providing any example pairs).

### 27. Describe how prompt engineering can impact GPT model performance.

**Prompt engineering** shapes the conditioning context provided to the model without altering its weights. Clear instructions, role assignment, structured formatting, few-shot exemplars, and step-by-step reasoning cues (*Chain-of-Thought*) dramatically reduce ambiguity, steer the model toward relevant knowledge subspaces, and improve accuracy, consistency, and reasoning depth.

### 28. Why are large datasets essential for training GPT models?

Large, diverse datasets expose GPT models to vast linguistic structures, factual knowledge, reasoning patterns, and domain contexts. In accordance with neural scaling laws, scaling both model parameters and training tokens prevents overfitting, enables emergent in-context learning abilities, and improves generalization across unseen tasks.

### 29. What are potential ethical concerns with GPT models?

- **Misinformation & Deepfakes:** Automated generation of persuasive fake news, phishing, or impersonation content.
- **Algorithmic Bias:** Reproducing racial, gender, or cultural stereotypes found in internet training data.
- **Copyright & Attribution:** Training on copyrighted works without explicit consent and generating derivative text.
- **Privacy Leakage:** Potential memorization and disclosure of sensitive PII present in training data.
- **Labor & Environmental Impact:** Job displacement in content-heavy roles and high carbon/water footprints of large-scale training.

### 30. How does the attention mechanism contribute to GPT's ability to handle long-range dependencies?

Causal self-attention computes direct pairwise dot-product interactions between the current token's Query vector and the Key vectors of *every* prior token in the context window. Unlike RNNs, information does not need to survive hundreds of recurrent hidden-state updates; a token at position 2,000 can directly access information from position 5 in a single computational hop.

### 31. What are some limitations of GPT models for real-world applications?

- **Hallucinations:** Confidently fabricating citations, facts, or code APIs.
- **Knowledge Cutoff:** Static parametric knowledge limited to the pre-training cutoff unless augmented with RAG or web search.
- **Finite Context Window:** Degradation in recall over very long contexts (*"lost in the middle"*) and quadratic attention compute costs.
- **Inference Latency & Cost:** Autoregressive token-by-token generation with large models incurs high compute and latency overhead.

### 32. How can GPT models be adapted for domain-specific text generation?

1. **Supervised Fine-Tuning (SFT) / PEFT (LoRA/QLoRA):** Train on curated domain documents or prompt-response pairs.
2. **Retrieval-Augmented Generation (RAG):** Connect GPT to an external domain vector database so it grounds answers in retrieved documents.
3. **Continued Pre-training:** Further pre-train on unlabeled domain corpora (e.g., legal, biomedical, financial text) to adapt vocabulary and domain priors.
4. **Prompt Engineering & Few-Shot In-Context Learning:** Provide domain rules and exemplars directly in the system/user prompt.

### 33. What are some common metrics for evaluating text generation quality?

- **Perplexity (PPL):** Measures how well a language model predicts a sample (lower is better).
- **BLEU & ROUGE:** N-gram precision (BLEU, common in translation) and recall/overlap (ROUGE, common in summarization) against reference texts.
- **METEOR & BERTScore:** Evaluate semantic similarity using synonyms, stemming, or contextual embeddings.
- **LLM-as-a-Judge & Human Evaluation:** Assess fluency, coherence, helpfulness, and factual faithfulness.

### 34. Explain the difference between deterministic and probabilistic text generation.

- **Deterministic Decoding:** Always selects the single highest-probability token (or path) at each step—such as **Greedy Search** or **Beam Search**. Given the same prompt, it always produces the exact same output, which is factual and consistent but often repetitive or bland.
- **Probabilistic (Stochastic) Decoding:** Samples the next token randomly according to the predicted probability distribution—such as **Temperature Sampling**, **Top-$k$**, or **Top-$p$ (Nucleus) Sampling**. This introduces controlled randomness, producing more creative, varied, and human-like text.

### 35. How does beam search improve text generation in language models?

Instead of greedily committing to only the single highest-probability token at each step (which can lead to suboptimal overall sequences), **beam search** maintains the top $B$ (beam width) most probable partial sequences simultaneously at every decoding step. By evaluating cumulative sequence log-probabilities across multiple hypotheses, it discovers more fluent, coherent, and globally high-probability outputs.

```python
# Practical 1: Write a code to generate a random sentence using probabilistic modeling (Markov Chain).
# Use the sentence "The cat is on the mat" as an example

import random
from collections import defaultdict

corpus = "The cat is on the mat"
words = corpus.split()

# Build first-order Markov transition dictionary
markov_chain = defaultdict(list)
for current_word, next_word in zip(words[:-1], words[1:]):
    markov_chain[current_word].append(next_word)

def generate_sentence(chain, start_word="The", max_length=10):
    sentence = [start_word]
    current = start_word
    for _ in range(max_length - 1):
        next_words = chain.get(current)
        if not next_words:
            break
        current = random.choice(next_words)
        sentence.append(current)
    return " ".join(sentence)

random.seed(42)
print("Transition table:", dict(markov_chain))
print("Generated sentence:", generate_sentence(markov_chain, start_word="The", max_length=6))
```

```python
# Practical 2: Build a simple Autoencoder model using Keras to learn a compressed representation of a given sentence.
# Use a dataset of your choice.

import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Embedding, LSTM, RepeatVector, TimeDistributed, Dense

sentences = [
    "the cat is on the mat",
    "the dog is in the yard",
    "artificial intelligence builds smart systems",
    "deep learning models learn representations",
    "autoencoders compress and reconstruct data",
    "neural networks process sequential text"
]

tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)
vocab_size = len(tokenizer.word_index) + 1
sequences = tokenizer.texts_to_sequences(sentences)
max_len = max(len(seq) for seq in sequences)
padded_seqs = pad_sequences(sequences, maxlen=max_len, padding="post")

# Autoencoder architecture
latent_dim = 8
inputs = Input(shape=(max_len,))
embedded = Embedding(input_dim=vocab_size, output_dim=16, mask_zero=False)(inputs)
encoded = LSTM(latent_dim, name="bottleneck_encoder")(embedded)

decoded = RepeatVector(max_len)(encoded)
decoded = LSTM(16, return_sequences=True)(decoded)
outputs = TimeDistributed(Dense(vocab_size, activation="softmax"))(decoded)

autoencoder = Model(inputs, outputs)
encoder = Model(inputs, encoded)

autoencoder.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
autoencoder.fit(padded_seqs, np.expand_dims(padded_seqs, -1), epochs=200, verbose=0)

# Extract compressed latent vector and reconstruct sample sentence
sample = padded_seqs[:1]
compressed_repr = encoder.predict(sample, verbose=0)
reconstructed_ids = np.argmax(autoencoder.predict(sample, verbose=0), axis=-1)[0]
reverse_word_map = {idx: word for word, idx in tokenizer.word_index.items()}
reconstructed_sentence = " ".join(reverse_word_map.get(i, "") for i in reconstructed_ids if i > 0)

print("Original sentence:", sentences[0])
print("Compressed latent representation (shape {}):\n".format(compressed_repr.shape), compressed_repr[0])
print("Reconstructed sentence:", reconstructed_sentence)
```

```python
# Practical 3: Use the Hugging Face transformers library to fine-tune a pre-trained GPT-2 model on a custom text data and generate text.

import torch
from torch.utils.data import Dataset
from transformers import GPT2Tokenizer, GPT2LMHeadModel, Trainer, TrainingArguments, DataCollatorForLanguageModeling

custom_texts = [
    "Generative AI creates original text, code, and images from learned data distributions.",
    "Transformers rely on self-attention mechanisms to model long-range dependencies in text.",
    "Fine-tuning adapts a pre-trained language model to specialized domain knowledge."
] * 10

model_name = "gpt2"
tokenizer = GPT2Tokenizer.from_pretrained(model_name)
tokenizer.pad_token = tokenizer.eos_token
model = GPT2LMHeadModel.from_pretrained(model_name)

class CustomTextDataset(Dataset):
    def __init__(self, texts, tokenizer, max_length=64):
        self.encodings = tokenizer(texts, truncation=True, max_length=max_length, padding="max_length", return_tensors="pt")

    def __len__(self):
        return self.encodings["input_ids"].size(0)

    def __getitem__(self, idx):
        return {
            "input_ids": self.encodings["input_ids"][idx],
            "attention_mask": self.encodings["attention_mask"][idx],
            "labels": self.encodings["input_ids"][idx].clone()
        }

train_dataset = CustomTextDataset(custom_texts, tokenizer)
data_collator = DataCollatorForLanguageModeling(tokenizer=tokenizer, mlm=False)

training_args = TrainingArguments(
    output_dir="./gpt2-custom-finetuned",
    num_train_epochs=3,
    per_device_train_batch_size=4,
    logging_steps=10,
    save_strategy="no",
    report_to="none"
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    data_collator=data_collator
)

trainer.train()

# Generate text using the fine-tuned model
prompt = "Generative AI creates"
inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
outputs = model.generate(**inputs, max_new_tokens=25, do_sample=True, top_p=0.9, pad_token_id=tokenizer.eos_token_id)
print("Fine-tuned GPT-2 output:", tokenizer.decode(outputs[0], skip_special_tokens=True))
```

```python
# Practical 4: Implement a text generation model using a simple Recurrent Neural Network (RNN) in Keras.
# Train the model on a custom data and generate a word.

import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

corpus = [
    "the cat sat on the warm mat",
    "the dog played in the green park",
    "machine learning models predict the next word",
    "simple recurrent neural networks process sequences"
]

tokenizer = Tokenizer()
tokenizer.fit_on_texts(corpus)
total_words = len(tokenizer.word_index) + 1

input_sequences = []
for line in corpus:
    token_list = tokenizer.texts_to_sequences([line])[0]
    for i in range(1, len(token_list)):
        input_sequences.append(token_list[:i + 1])

max_seq_len = max(len(seq) for seq in input_sequences)
input_sequences = pad_sequences(input_sequences, maxlen=max_seq_len, padding="pre")

X, y = input_sequences[:, :-1], input_sequences[:, -1]
y = tf.keras.utils.to_categorical(y, num_classes=total_words)

model = Sequential([
    Embedding(total_words, 16, input_length=max_seq_len - 1),
    SimpleRNN(32),
    Dense(total_words, activation="softmax")
])

model.compile(loss="categorical_crossentropy", optimizer="adam", metrics=["accuracy"])
model.fit(X, y, epochs=250, verbose=0)

seed_text = "the cat sat on the"
token_list = pad_sequences([tokenizer.texts_to_sequences([seed_text])[0]], maxlen=max_seq_len - 1, padding="pre")
predicted_id = np.argmax(model.predict(token_list, verbose=0), axis=-1)[0]
predicted_word = tokenizer.index_word.get(predicted_id, "")

print(f"Seed text: '{seed_text}' -> Next predicted word: '{predicted_word}'")
```

```python
# Practical 5: Write a program to generate a sequence of text using an LSTM-based model in TensorFlow, trained on a custom data of sentences.

import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

sentences = [
    "deep learning enables natural language understanding",
    "long short term memory networks capture sequence patterns",
    "tensorflow makes building neural networks straightforward",
    "language models generate coherent sequences of text"
]

tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)
vocab_size = len(tokenizer.word_index) + 1

sequences = []
for sentence in sentences:
    tokens = tokenizer.texts_to_sequences([sentence])[0]
    for i in range(1, len(tokens)):
        sequences.append(tokens[:i + 1])

max_len = max(len(s) for s in sequences)
sequences = pad_sequences(sequences, maxlen=max_len, padding="pre")
X, y = sequences[:, :-1], tf.keras.utils.to_categorical(sequences[:, -1], num_classes=vocab_size)

model = Sequential([
    Embedding(vocab_size, 24, input_length=max_len - 1),
    LSTM(64),
    Dense(vocab_size, activation="softmax")
])

model.compile(loss="categorical_crossentropy", optimizer="adam", metrics=["accuracy"])
model.fit(X, y, epochs=250, verbose=0)

def generate_sequence(seed_text, num_words=5):
    result = seed_text
    for _ in range(num_words):
        encoded = pad_sequences([tokenizer.texts_to_sequences([result])[0]], maxlen=max_len - 1, padding="pre")
        next_id = np.argmax(model.predict(encoded, verbose=0), axis=-1)[0]
        next_word = tokenizer.index_word.get(next_id, "")
        if not next_word:
            break
        result += " " + next_word
    return result

print("Generated sequence:", generate_sequence("deep learning enables", num_words=4))
```

```python
# Practical 6: Build a program that uses GPT-2 from Hugging Face to generate a story based on a custom prompt.

from transformers import GPT2Tokenizer, GPT2LMHeadModel, set_seed

set_seed(42)
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")

prompt = "Once upon a time in a quiet coastal town, an old lighthouse keeper discovered a glowing map inside a bottle."
inputs = tokenizer(prompt, return_tensors="pt")

story_ids = model.generate(
    **inputs,
    max_new_tokens=100,
    do_sample=True,
    top_k=50,
    top_p=0.95,
    temperature=0.8,
    no_repeat_ngram_size=2,
    pad_token_id=tokenizer.eos_token_id
)

story = tokenizer.decode(story_ids[0], skip_special_tokens=True)
print("Generated Story:\n", story)
```

```python
# Practical 7: Write a code to implement a simple text generation model using a GRU-based architecture in Keras.

import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, GRU, Dense

corpus = [
    "gated recurrent units are efficient for sequence modeling",
    "gru models use update and reset gates to control memory",
    "keras provides clean abstractions for deep learning"
]

tokenizer = Tokenizer()
tokenizer.fit_on_texts(corpus)
vocab_size = len(tokenizer.word_index) + 1

sequences = []
for line in corpus:
    tokens = tokenizer.texts_to_sequences([line])[0]
    for i in range(1, len(tokens)):
        sequences.append(tokens[:i + 1])

max_len = max(len(s) for s in sequences)
sequences = pad_sequences(sequences, maxlen=max_len, padding="pre")
X, y = sequences[:, :-1], tf.keras.utils.to_categorical(sequences[:, -1], num_classes=vocab_size)

gru_model = Sequential([
    Embedding(vocab_size, 24, input_length=max_len - 1),
    GRU(64),
    Dense(vocab_size, activation="softmax")
])

gru_model.compile(loss="categorical_crossentropy", optimizer="adam", metrics=["accuracy"])
gru_model.fit(X, y, epochs=250, verbose=0)

seed = "gated recurrent units"
for _ in range(5):
    encoded = pad_sequences([tokenizer.texts_to_sequences([seed])[0]], maxlen=max_len - 1, padding="pre")
    next_id = np.argmax(gru_model.predict(encoded, verbose=0), axis=-1)[0]
    seed += " " + tokenizer.index_word.get(next_id, "")

print("GRU Generated Text:", seed)
```

```python
# Practical 8: Create a script to implement GPT-2-based text generation with beam search decoding to generate text.

from transformers import GPT2Tokenizer, GPT2LMHeadModel

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")

prompt = "Artificial intelligence in healthcare is transforming"
inputs = tokenizer(prompt, return_tensors="pt")

beam_outputs = model.generate(
    **inputs,
    max_new_tokens=45,
    num_beams=5,
    no_repeat_ngram_size=2,
    early_stopping=True,
    pad_token_id=tokenizer.eos_token_id
)

print("Beam Search Output:\n", tokenizer.decode(beam_outputs[0], skip_special_tokens=True))
```

```python
# Practical 9: Implement a text generation script using GPT-2 with a custom temperature setting for diversity in output text.

from transformers import GPT2Tokenizer, GPT2LMHeadModel, set_seed

set_seed(42)
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")

prompt = "In the future, space exploration will allow humanity to"
inputs = tokenizer(prompt, return_tensors="pt")

custom_temperature = 0.85
outputs = model.generate(
    **inputs,
    max_new_tokens=45,
    do_sample=True,
    temperature=custom_temperature,
    top_k=50,
    pad_token_id=tokenizer.eos_token_id
)

print(f"Generated Text (temperature={custom_temperature}):\n", tokenizer.decode(outputs[0], skip_special_tokens=True))
```

```python
# Practical 10: Create a script to implement temperature sampling with GPT-2, experimenting with different values to generate creative text.

from transformers import GPT2Tokenizer, GPT2LMHeadModel, set_seed

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")

prompt = "The robot looked up at the starry night sky and whispered,"
inputs = tokenizer(prompt, return_tensors="pt")

temperatures = [0.3, 0.7, 1.2]

for temp in temperatures:
    set_seed(42)
    output_ids = model.generate(
        **inputs,
        max_new_tokens=40,
        do_sample=True,
        temperature=temp,
        top_k=50,
        pad_token_id=tokenizer.eos_token_id
    )
    generated_text = tokenizer.decode(output_ids[0], skip_special_tokens=True)
    print(f"--- Temperature: {temp} ---")
    print(generated_text, "\n")
```

```python
# Practical 11: Implement a simple LSTM-based text generation model from scratch using Keras and train it on a custom data.

import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

# Character-level LSTM trained from scratch on custom data
text = (
    "generative artificial intelligence learns patterns from data to create new text. "
    "recurrent neural networks like lstm maintain cell state across time steps."
)

chars = sorted(list(set(text)))
char_to_idx = {c: i for i, c in enumerate(chars)}
idx_to_char = {i: c for i, c in enumerate(chars)}
vocab_size = len(chars)

seq_length = 25
step = 3
sequences = []
next_chars = []

for i in range(0, len(text) - seq_length, step):
    sequences.append(text[i: i + seq_length])
    next_chars.append(text[i + seq_length])

# One-hot encode sequences from scratch
X = np.zeros((len(sequences), seq_length, vocab_size), dtype=np.float32)
y = np.zeros((len(sequences), vocab_size), dtype=np.float32)

for i, seq in enumerate(sequences):
    for t, char in enumerate(seq):
        X[i, t, char_to_idx[char]] = 1.0
    y[i, char_to_idx[next_chars[i]]] = 1.0

model = Sequential([
    LSTM(64, input_shape=(seq_length, vocab_size)),
    Dense(vocab_size, activation="softmax")
])

model.compile(loss="categorical_crossentropy", optimizer="adam")
model.fit(X, y, batch_size=16, epochs=150, verbose=0)

# Generate characters starting from a seed string
seed = text[:seq_length]
generated = seed
for _ in range(50):
    x_pred = np.zeros((1, seq_length, vocab_size), dtype=np.float32)
    for t, char in enumerate(seed):
        x_pred[0, t, char_to_idx[char]] = 1.0
    preds = model.predict(x_pred, verbose=0)[0]
    next_char = idx_to_char[np.argmax(preds)]
    generated += next_char
    seed = seed[1:] + next_char

print("Character-level LSTM generated text:\n", generated)
```

```python
# Practical 12: How can you implement text generation using a simple custom attention-based architecture?

import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Embedding, Layer, Dense, GlobalAveragePooling1D

# Custom Scaled Dot-Product Self-Attention Layer
class SimpleSelfAttention(Layer):
    def __init__(self, d_model):
        super().__init__()
        self.d_model = d_model
        self.wq = Dense(d_model)
        self.wk = Dense(d_model)
        self.wv = Dense(d_model)

    def call(self, inputs):
        q = self.wq(inputs)
        k = self.wk(inputs)
        v = self.wv(inputs)
        scores = tf.matmul(q, k, transpose_b=True) / tf.math.sqrt(tf.cast(self.d_model, tf.float32))
        weights = tf.nn.softmax(scores, axis=-1)
        return tf.matmul(weights, v)

corpus = [
    "attention mechanisms allow models to focus on relevant words",
    "self attention computes relationships across all sequence tokens",
    "transformers use attention for fast parallel text generation"
]

tokenizer = Tokenizer()
tokenizer.fit_on_texts(corpus)
vocab_size = len(tokenizer.word_index) + 1

sequences = []
for line in corpus:
    tokens = tokenizer.texts_to_sequences([line])[0]
    for i in range(1, len(tokens)):
        sequences.append(tokens[:i + 1])

max_len = max(len(s) for s in sequences)
sequences = pad_sequences(sequences, maxlen=max_len, padding="pre")
X, y = sequences[:, :-1], tf.keras.utils.to_categorical(sequences[:, -1], num_classes=vocab_size)

# Build model with custom attention layer
inputs = Input(shape=(max_len - 1,))
x = Embedding(vocab_size, 32)(inputs)
x = SimpleSelfAttention(d_model=32)(x)
x = GlobalAveragePooling1D()(x)
outputs = Dense(vocab_size, activation="softmax")(x)

attention_model = Model(inputs, outputs)
attention_model.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])
attention_model.fit(X, y, epochs=250, verbose=0)

seed = "attention mechanisms allow"
for _ in range(5):
    encoded = pad_sequences([tokenizer.texts_to_sequences([seed])[0]], maxlen=max_len - 1, padding="pre")
    next_id = np.argmax(attention_model.predict(encoded, verbose=0), axis=-1)[0]
    seed += " " + tokenizer.index_word.get(next_id, "")

print("Custom Attention Model Output:", seed)
```
